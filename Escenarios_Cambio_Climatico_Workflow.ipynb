# Análisis de escenarios de cambio climático para recursos hidrotermales

En este proyecto voy a analizar cómo pueden evolucionar distintas variables climáticas de interés para la gestión de recursos hidrotermales bajo diferentes escenarios de cambio climático.

El trabajo se desarrolla en el marco del proyecto ThermEcoWat y se centra en tres variables principales: precipitación, temperatura y evapotranspiración potencial. Para ello se utilizan datos climáticos observados y proyecciones regionalizadas obtenidas a partir de modelos climáticos de la iniciativa EURO-CORDEX. 

El objetivo final es generar una serie de mapas comparables entre sí que permitan identificar tendencias futuras y evaluar posibles impactos sobre los territorios estudiados.

# Objetivos

Los objetivos principales de este análisis son:

- Obtener información climática histórica para los territorios de estudio.
- Analizar las proyecciones climáticas disponibles para diferentes escenarios de cambio climático.
- Comparar los valores históricos con los escenarios futuros.
- Generar cartografía temática de precipitación, temperatura y evapotranspiración potencial.
- Evaluar visualmente las diferencias entre escenarios y periodos temporales.

Todo el proceso se ha realizado utilizando herramientas SIG y datos climáticos abiertos procedentes de organismos oficiales.



# Preparación de la información climática

Una vez identificadas las fuentes de información disponibles, comienzo la descarga de los datos necesarios para el análisis.

Para el caso de España, las proyecciones climáticas son proporcionadas por la Agencia Estatal de Meteorología (AEMET), que genera y revisa periódicamente escenarios climáticos regionalizados a partir de modelos globales incluidos en el Quinto Informe de Evaluación del IPCC (AR5) y del proyecto CORDEX.

Durante el trabajo utilizo dos fuentes principales de datos:

- La plataforma AEMET, que proporciona datos climáticos brutos con resolución temporal mensual.
- El visor ADAPTECCA, que ofrece resultados ya procesados y agregados para periodos de 30 años.

Dado que el objetivo final consiste en generar cartografía climática comparativa para distintos escenarios de cambio climático, la mayor parte de la información utilizada procede del visor ADAPTECCA.

Todos los datos descargados se incorporan posteriormente a un proyecto QGIS, donde son organizados según la variable analizada, el escenario de emisiones y el horizonte temporal correspondiente.

Esta organización resulta especialmente importante cuando se trabaja simultáneamente con decenas de capas ráster procedentes de diferentes modelos climáticos regionales.

# Selección de escenarios y periodos temporales

El visor ADAPTECCA proporciona información climática para diferentes escenarios futuros de emisiones de gases de efecto invernadero.

En este análisis utilizo los dos escenarios disponibles:

- RCP4.5, considerado un escenario de emisiones intermedias.
- RCP8.5, que representa un escenario de emisiones elevadas.

Además, cada escenario se analiza para dos horizontes temporales de interés:

- Medio plazo (2041–2070).
- Largo plazo (2071–2100).

Como periodo de referencia utilizo el escenario histórico correspondiente al intervalo 1971–2000.

Esta línea base permite calcular posteriormente las anomalías climáticas y evaluar cómo podrían evolucionar las distintas variables respecto a las condiciones observadas durante el periodo de referencia.

ADAPTECCA ofrece dos tipos de proyecciones climáticas:

- Proyecciones puntuales obtenidas mediante downscaling estadístico.
- Proyecciones en malla obtenidas mediante downscaling dinámico a partir de EURO-CORDEX.

# Procesamiento raster y generación de anomalías

Una vez descargadas todas las capas climáticas disponibles en ADAPTECCA, el siguiente paso consiste en procesarlas para obtener una representación única para cada variable, escenario y horizonte temporal.

Los datos originales incluyen múltiples simulaciones procedentes de diferentes combinaciones de modelos climáticos globales (GCM) y modelos climáticos regionales (RCM). Cada uno de estos modelos genera resultados ligeramente distintos, por lo que trabajar únicamente con una simulación individual podría introducir sesgos asociados a las características propias del modelo seleccionado.

Por este motivo se optó por utilizar un enfoque multimodelo, ampliamente empleado en estudios climáticos, que consiste en combinar varias simulaciones para obtener una estimación más robusta del comportamiento futuro de las variables analizadas.


## Tratamiento de la incertidumbre

La incertidumbre presente en las proyecciones climáticas proviene principalmente de tres fuentes:

### Incertidumbre asociada al escenario de emisiones

Depende de la evolución futura de factores socioeconómicos como el crecimiento demográfico, los patrones de consumo energético o las políticas de mitigación.

Su influencia suele ser relativamente reducida durante las primeras décadas del siglo, pero aumenta progresivamente hasta convertirse en una de las principales fuentes de incertidumbre a finales del siglo XXI.

### Incertidumbre asociada a la modelización climática

Diferentes modelos climáticos pueden producir respuestas distintas frente a un mismo escenario de emisiones.

Esta componente suele alcanzar su máxima importancia en horizontes temporales intermedios.

### Variabilidad natural del clima

Corresponde a la variabilidad interna propia del sistema climático.

Su peso relativo suele ser más importante en los primeros años de las proyecciones y disminuye a medida que aumenta la influencia de los escenarios de emisiones y de las diferencias entre modelos.

Por este motivo resulta necesario analizar más de un escenario climático y trabajar con varios modelos para caracterizar adecuadamente la incertidumbre asociada a las proyecciones.

## Selección del conjunto de modelos

Para minimizar este problema selecciono un subconjunto representativo siguiendo las recomendaciones planteadas por Weigel et al. (2010), evitando duplicidades entre familias de modelos y buscando una representación equilibrada de la incertidumbre disponible.

# Generación de la temperatura media anual

Durante la recopilación de datos se detectó que ADAPTECCA no proporcionaba directamente mapas de temperatura media anual.

Únicamente estaban disponibles las variables:

- Temperatura mínima.
- Temperatura máxima.

Por este motivo fue necesario generar una capa adicional de temperatura media anual utilizando operaciones raster.

La temperatura media se calculó para cada celda aplicando la expresión:

Tmedia = (Tmáxima + Tmínima) / 2

Este procedimiento se aplicó tanto al periodo histórico como a todos los escenarios futuros disponibles.

Una vez completado el proceso se obtuvieron nuevas capas de temperatura media anual que permitieron realizar comparaciones homogéneas entre todos los periodos analizados.

![06_Escenarios_Climate_Change](./images/Mean_Annual_Temperature.png)

# Cálculo de anomalías absolutas

Tras generar las capas climáticas multimodelo, el siguiente paso consistió en cuantificar los cambios respecto al periodo histórico de referencia (1971-2000).

Para ello se calcularon anomalías absolutas, que representan la diferencia directa entre el escenario futuro y el escenario histórico.

La expresión utilizada fue:

Anomalía absoluta = Escenario futuro - Escenario histórico

Este cálculo se realizó para:

- Temperatura media.
- Precipitación.
- Evapotranspiración potencial.

Las anomalías absolutas permiten identificar de forma inmediata cuánto aumenta o disminuye una variable respecto a las condiciones históricas observadas.

En el caso de la temperatura, los resultados obtenidos se expresan en grados Celsius, mientras que para precipitación y evapotranspiración se mantienen las unidades originales de cada variable.

# Cálculo de anomalías relativas

Además de las diferencias absolutas, también se calcularon anomalías relativas para facilitar la comparación entre variables con magnitudes diferentes.

Las anomalías relativas expresan el cambio porcentual respecto al valor histórico y se calcularon mediante la siguiente expresión:

Anomalía relativa (%) =
((Escenario futuro - Histórico) / Histórico) × 100

Este indicador resulta especialmente útil para interpretar la evolución de la precipitación y de la evapotranspiración potencial.

Mientras que una diferencia absoluta puede resultar difícil de contextualizar, la representación porcentual permite identificar rápidamente la magnitud real del cambio proyectado.

Por ejemplo, una reducción del 10 % de la precipitación anual puede interpretarse de forma más intuitiva que una disminución expresada únicamente en milímetros por año.

# Evolución de la temperatura

La temperatura media anual observada durante el periodo histórico (1971–2000) es de aproximadamente 14.37 °C en el entorno de Caldes de Montbui.

Tras generar los escenarios futuros se observa una tendencia clara y consistente al aumento de la temperatura en todos los modelos analizados.

En el escenario RCP4.5 se proyecta un incremento aproximado de 2.28 °C al final de siglo.

Sin embargo, bajo el escenario RCP8.5 el aumento alcanza valores cercanos a 4.25 °C.

Estas diferencias ponen de manifiesto la influencia que tienen las emisiones futuras de gases de efecto invernadero sobre la evolución climática del territorio.

Para representar gráficamente este comportamiento se generaron capas de anomalías térmicas que muestran la diferencia entre cada escenario futuro y el periodo histórico de referencia.

# Evolución de la precipitación

La precipitación presenta un comportamiento más complejo y variable que la temperatura.

Durante el periodo histórico, la precipitación media anual en el área de estudio es de aproximadamente 675 mm/año.

Tras comparar los diferentes escenarios disponibles se observa una elevada variabilidad entre modelos, lo que dificulta identificar una tendencia única para todo el periodo analizado.

Los escenarios RCP4.5 y algunos escenarios RCP8.5 para mediados de siglo sugieren ligeros incrementos de precipitación en buena parte de la región mediterránea.

Sin embargo, los resultados cambian hacia finales de siglo.

Bajo el escenario RCP8.5 para el periodo 2071–2100 se proyecta una disminución generalizada de las precipitaciones, alcanzando en Caldes de Montbui valores cercanos al 10 % de reducción respecto al periodo histórico.

Los mapas de anomalía generados permiten visualizar espacialmente estas diferencias y valorar su posible impacto sobre la disponibilidad futura de recursos hídricos.

![06_Escenarios_Climate_Change](./images/Mean_Annual_Precipitation.png)

# Evolución de la evapotranspiración potencial

Además de la temperatura y la precipitación, se analiza la evapotranspiración potencial debido a su estrecha relación con los procesos hidrológicos y el balance hídrico de los sistemas naturales.

Para ello se incorporan las capas históricas observadas y las proyecciones futuras correspondientes a cada escenario climático.

Siguiendo la misma metodología aplicada al resto de variables, se calculan las anomalías respecto al periodo histórico con el fin de identificar incrementos o disminuciones futuras.

Los resultados muestran una tendencia general al aumento de la evapotranspiración potencial, especialmente en aquellos escenarios donde el incremento térmico es más acusado.

Esta situación podría traducirse en una mayor demanda evaporativa y en una reducción de los recursos hídricos disponibles durante determinadas épocas del año.

# Elaboración de la cartografía final

Una vez generadas todas las capas de anomalías y los distintos escenarios climáticos, comienzo el diseño de los mapas finales mediante el gestor de composiciones de QGIS.

Para mantener una representación homogénea entre variables, todos los mapas comparten una estructura similar:

- Panel de referencia con la situación histórica.
- Mapas para el horizonte medio (2041–2070).
- Mapas para el horizonte largo (2071–2100).
- Comparación entre escenarios RCP4.5 y RCP8.5.
- Localización precisa del piloto de Caldes de Montbui.

Además de los mapas generales para toda la Península Ibérica, incorporo ampliaciones específicas del área de estudio que permiten visualizar con mayor detalle los cambios proyectados.

En cada composición se incluyen leyendas normalizadas, escalas de color coherentes entre escenarios y etiquetas numéricas que muestran el cambio estimado para el piloto analizado.

El resultado final es una serie de mapas comparables entre sí que permiten interpretar rápidamente la evolución futura de la temperatura, la precipitación y la evapotranspiración potencial bajo diferentes escenarios de cambio climático.

# Conclusiones

El análisis realizado muestra que la temperatura es la variable que presenta una señal de cambio más robusta y consistente entre escenarios.

Mientras que el escenario RCP4.5 proyecta incrementos moderados de temperatura, el escenario RCP8.5 evidencia un calentamiento mucho más acusado hacia finales de siglo.

La precipitación presenta una mayor incertidumbre, con incrementos y disminuciones que dependen del horizonte temporal y del escenario considerado. No obstante, los resultados apuntan hacia una reducción significativa de las precipitaciones bajo el escenario de emisiones más elevadas al final del siglo.

Por su parte, la evapotranspiración potencial muestra una tendencia creciente coherente con el aumento proyectado de la temperatura.

En conjunto, los resultados sugieren una posible intensificación de las condiciones de estrés hídrico en el territorio estudiado, aspecto especialmente relevante para la planificación y gestión futura de los recursos hidrotermales asociados al proyecto ThermEcoWat.
